3. UPDATE

3.1 Dropping Rows

```python
# Drop rows where Total day minutes < 50
df_condition_drop = df[df["Total day minutes"] >= 50]
df_condition_drop.head()
```

3.2 Dropping Columns

```python
# 3.2 Dropping Columns
df_drop_cols = df.drop(columns=["Area code", "Voice mail plan"])
df_drop_cols.head()
```

3.3 Creating Calculated Columns

```python
# 3.3 Creating Calculated Columns
# Example: Total charges = day + evening + night + intl charges
df["Total Charges"] = (
    df["Total day charge"] +
    df["Total eve charge"] +
    df["Total night charge"] +
    df["Total intl charge"]
)

df[["Total Charges"]].head()
```

3.6 Condition-Based Updating Using Apply

```python
# 3.6 Condition-based updating using apply
# Example: Label heavy users based on Total day minutes
def usage_label(row):
    if row["Total day minutes"] > 250:
        return "High"
    elif row["Total day minutes"] > 150:
        return "Medium"
    else:
        return "Low"

df["Usage Category"] = df.apply(usage_label, axis=1)
df[["Total day minutes", "Usage Category"]].head()
```

4. DELETE / OUTPUT

```python
# 4.1 Output to CSV
df.to_csv("cleaned_telco.csv", index=False)
```

4.2 Output to JSON

```python
# 4.2 Output to JSON
df.to_json("telco.json", orient="records")
```
